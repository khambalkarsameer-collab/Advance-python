```python
import numpy as np
students=np.array([
    [85,90,78],
    [60,55,70],
    [95,92,88],
    [40,45,50],
    [72,68,75]
])
students
```

```python
avg_marks=np.mean(students,axis=1)
print("Avg marks of students:")
print(avg_marks)
```

```python
avg_marks=np.mean(students,axis=0)
print("Avg marks of students:")
print(avg_marks)
```

```python
topper=np.argmax(avg_marks)
print("Topper is:",topper)
```

```python
failed_students = np.where(avg_marks<80)
print("Failed students are:",failed_students[0]+1)
```

```python
# performance report
for i, avg in enumerate(avg_marks):
  status="pass"  if avg>50 else "fail"
  print(f"Student {i+1} Avg Marks: {avg}")
```

```python
print(f"{'Student':<15}{'Avg Marks':<15}{'Status':<10}")

for i, avg in enumerate(avg_marks):
    status = "Pass" if avg > 50 else "Fail"
    print(f"{'Student ' + str(i+1):<15}{avg:<15}{status:<10}")
```

```python
import matplotlib.pyplot as plt
```

```python
x=[1,2,3,4]
y=[2,4,6,8]
plt.plot(x,y)
plt.title('Line Plot')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.grid(True)
plt.show()
```

```python
x=[-1,-2,0,1,2]
y=[0.25,0.5,1,2,4]
plt.plot(x,y)
```

```python
months=['jan','feb','march','april']
sales=[2250,300,400,350]
plt.bar(months,sales)
plt.title('monthly sales')
plt.xlabel('Months')
plt.ylabel('Sales')
plt.show()
```

```python
plt.barh(months,sales,color='green')
plt.title('monthly sales')
plt.xlabel('sales')
plt.ylabel('months')
plt.show()
```

```python
height=[150,160,170,180,190,200]
weight=[50,60,70,80,90,110]
plt.scatter(height,weight)
plt.title('height vs weight')
plt.xlabel('height(cm)')
plt.ylabel('weight(kg)')
plt.show()
```

```python
brands=['brand a','brand b','brand c']
shares =[45,30,25]
plt.pie(shares,labels=brands,autopct='%1.1f%%')
plt.title('Market Share')
plt.axis('equal')
plt.show()
```

```python
x=[1,2,3,4]
y=[2,4,6,8]
y1=[3,6,9,12]
plt.plot(x,y)
plt.title('multiple lines')
plt.plot(x,y1,label='square')
plt.plot(x,y,label='linear')
plt.legend()
plt.grid(True)
plt.show()
#
```

```python
fig,axs=plt.subplots(2,2)

axs[0,0].plot(x,y)
axs[0,0].set_title('linear')

axs[0,1].plot(months,sales)
axs[0,1].set_title('sales')

axs[1,0].plot(height,weight)
axs[1,0].set_title('scatter')

axs[1,1].pie(shares,labels=brands,autopct='%1.1f%%')
axs[1,1].set_title('pie')

plt.tight_layout()
plt.show()
```
